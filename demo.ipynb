```python
from dice import Dice
from items import Weapon
from character import Character, Warrior, Mage
from battle import Battle

sword = Weapon("Sword", Dice(8))
hero = Warrior("Brienne", weapon=sword)
enemy = Mage("Merlin")

winner = Battle(hero, enemy).simulate_battle()
print(winner, "| hero HP:", hero.hp, "| enemy HP:", enemy.hp)
```

```python
wins = {"Warrior": 0, "Mage": 0}
for _ in range(100):
    hero = Warrior("Brienne", weapon=Weapon("Sword", Dice(8)))
    enemy = Mage("Merlin")
    winner = Battle(hero, enemy).simulate_battle()
    wins[type(winner).__name__] += 1
print(wins)
```
